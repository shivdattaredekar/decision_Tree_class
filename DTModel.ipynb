```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn.model_selection import train_test_split
from sklearn import metrics
```

```python
df = pd.read_csv('D:\Datascience\Training\BIA\ML\Regression\Logistic\iris.csv')
df.head()
```

visualize the data set

```python
plt.figure(figsize=(12,6))

for x, y in df.groupby('species'):
    plt.scatter(y['sepal_length'], y['sepal_width'], label = x)

plt.xlabel('sepal_length')
plt.ylabel('sepal_width')

plt.legend(title = 'species')
plt.show()
```

```python
df.columns
```

```python
X = df.drop(columns={'species'})
y = df['species']
```

```python
x_train ,x_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.7
)
```

DT model

```python
model = DecisionTreeClassifier()
model.fit(x_train, y_train)
```

```python
import pydotplus
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# 1. Setup a matplotlib figure size
plt.figure(figsize=(15, 10))

# 2. Plot the tree directly
plot_tree(
    model,
    feature_names=X.columns,
    class_names=y.unique(),
    filled=True,
    rounded=True
)

# 3. Save it directly via matplotlib
plt.savefig("Iris_decision_tree.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
y_pred = model.predict(x_test)
```

```python
metrics.accuracy_score(y_test, y_pred)
metrics.precision_score(y_test, y_pred, average='weighted')
metrics.recall_score(y_test, y_pred, average='weighted')
metrics.confusion_matrix(y_test, y_pred)
```
